# Circuit 1 Qubit : Portes X, H et Mesure

## Objectifs
- Créer un circuit quantique à 1 qubit
- Appliquer les portes Pauli-X et Hadamard
- Mesurer le qubit
- Simuler avec AerSimulator et afficher l'histogramme

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram

print("Bibliothèques prêtes ")
```

## 1. Création du circuit

- **Porte X (Pauli-X)** : inverse l'état du qubit (agit comme un NOT).
- **Porte H (Hadamard)** : place le qubit en superposition.

État initial : |0⟩ → après X → |1⟩ → après H → (|0⟩ - |1⟩)/√2

```python

qc = QuantumCircuit(1, 1)

qc.x(0)         
qc.h(0)          

qc.measure(0, 0)

qc.draw('mpl')


sim = AerSimulator()


result = sim.run(qc, shots=1024).result()


counts = result.get_counts()
print(counts)


plot_histogram(counts)
qc = QuantumCircuit(1, 1)

plot_histogram(counts)
```

## Conclusion
Après la porte X, le qubit passe de |0⟩ à |1⟩.
Après la porte H, il est en superposition.
La mesure donne donc 0 ou 1 avec une probabilité de 50% chacun.
